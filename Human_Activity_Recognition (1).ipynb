```python
import itertools
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import numpy as np
plt.rcParams["font.family"] = 'DejaVu Sans'

def plot_confusion_matrix(cm, classes,
                          normalize=False,
                          title='Confusion matrix',
                          cmap=plt.cm.Blues):
    """
    This function prints and plots the confusion matrix.
    Normalization can be applied by setting `normalize=True`.
    """
    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    # The following lines were not indented properly and were outside the function definition
    # Indenting them to be part of the function
    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    fmt = '.2f' if normalize else 'd'
    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt),
                horizontalalignment="center",
                color="white" if cm[i, j] > thresh else "black")

    plt.tight_layout()
    plt.ylabel('True label')
    plt.xlabel('Predicted label')
```

```python
def _count_classes(y):
  return len(set([tuple(category) for category in y]))
```

```python
import os
import numpy as np
import pandas as pd

SIGNALS = ["body_acc_x", "body_acc_y", "body_acc_z"]  # Define the signals list

def _read_csv(filename):
    """Helper function to read a CSV file."""
    if not os.path.exists(filename):
        raise FileNotFoundError(f"File not found: {filename}")
    return pd.read_csv(filename, delim_whitespace=True, header=None)

def load_signals(subset):
    signals_data = []

    for signal in SIGNALS:
        # Construct file path correctly
        filename = f'/content/HAR/{subset}/Inertial Signals/{signal}_{subset}.txt'

        # Ensure the file exists
        if not os.path.exists(filename):
            print(f"Warning: Missing file {filename}, skipping.")
            continue  # Skip missing files instead of breaking

        signals_data.append(_read_csv(filename).values)

    if not signals_data:
        raise ValueError("No signal data loaded. Please check file paths.")

    return np.transpose(signals_data, (1, 2, 0))

def load_y(subset):
    # Construct correct file path
    filename = f'/content/HAR/{subset}/y_{subset}.txt'

    if not os.path.exists(filename):
        raise FileNotFoundError(f"File not found: {filename}")

    y = _read_csv(filename)[0]
    return y

def load_data():
    X_train, X_test = load_signals('train'), load_signals('test')
    y_train, y_test = load_y('train'), load_y('test')
    return y
```
